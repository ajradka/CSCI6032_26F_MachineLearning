## Problem 3: Optimal Classifier with Respect to Error

An optimal classifier does not necessarily achieve zero error.
Sometimes there is **inherent ambiguity** — the same feature vector
$x$ can occur for more than one class.  The optimal classifier
always chooses the class with the highest posterior
probability $P(\omega_i \mid x)$, which we obtain from the
class likelihoods and priors using Bayes' rule.

Consider a one-dimensional feature variable $x \in [0,1]$.
There are two classes, $\omega_1$ and $\omega_2$, and the prior
probabilities are equal:

$$
P(\omega_1) = P(\omega_2) = \tfrac{1}{2}.
$$

The class-conditional densities (likelihoods) $p(x \mid \omega_1)$ and
$p(x \mid \omega_2)$ vary across three regions:

$$
p(x \mid \omega_1) =
\begin{cases}
2, & 0 \le x < \tfrac{1}{3} \quad \text{(Region A)} \\
4 - 6x, & \tfrac{1}{3} \le x \le \tfrac{2}{3} \quad \text{(Region B)} \\
0, & \tfrac{2}{3} < x \le 1 \quad \text{(Region C)}
\end{cases}
\qquad
p(x \mid \omega_2) =
\begin{cases}
0, & 0 \le x < \tfrac{1}{3} \quad \text{(Region A)} \\
6x - 2, & \tfrac{1}{3} \le x \le \tfrac{2}{3} \quad \text{(Region B)} \\
2, & \tfrac{2}{3} < x \le 1 \quad \text{(Region C)}
\end{cases}
$$

Both densities are zero outside $[0,1]$.

**(a)** Verify that each likelihood integrates to 1.  Find the evidence $p(x)$ using the law of
total probability, and use Bayes' rule to find $P(\omega_1 \mid x)$ and $P(\omega_2 \mid x)$
in each region. (3 points)

*Likelihoods integrate to 1* (areas computed geometrically):

$p(x \mid \omega_1)$: Region A (rectangle) $= \frac{1}{3} \cdot 2 = \frac{2}{3}$, Region B (triangle) $= \frac{1}{2} \cdot \frac{1}{3} \cdot 2 = \frac{1}{3}$, Region C $= 0$

$$\int_0^1 p(x \mid \omega_1)\,dx = \frac{2}{3} + \frac{1}{3} + 0 = 1$$

$p(x \mid \omega_2)$: Region A $= 0$, Region B (triangle) $= \frac{1}{3}$, Region C (rectangle) $= \frac{2}{3}$

$$\int_0^1 p(x \mid \omega_2)\,dx = 0 + \frac{1}{3} + \frac{2}{3} = 1$$

*Evidence* (law of total probability):

$$p(x) = p(x \mid \omega_1)P(\omega_1) + p(x \mid \omega_2)P(\omega_2)$$

- Region A: $\frac{1}{2}(2) + \frac{1}{2}(0) = 1$
- Region B: $\frac{1}{2}(4 - 6x) + \frac{1}{2}(6x - 2) = 1$
- Region C: $\frac{1}{2}(0) + \frac{1}{2}(2) = 1$

So $p(x) = 1$ on $[0,1]$ (uniform).

*Posteriors* (Bayes' rule): $P(\omega_i \mid x) = \dfrac{p(x \mid \omega_i)P(\omega_i)}{p(x)} = \frac{1}{2}\,p(x \mid \omega_i)$

$$P(\omega_1 \mid x) = \begin{cases} 1, & 0 \le x < \frac{1}{3} \text{ (A)} \\ 2 - 3x, & \frac{1}{3} \le x \le \frac{2}{3} \text{ (B)} \\ 0, & \frac{2}{3} < x \le 1 \text{ (C)} \end{cases} \qquad P(\omega_2 \mid x) = \begin{cases} 0, & 0 \le x < \frac{1}{3} \text{ (A)} \\ 3x - 1, & \frac{1}{3} \le x \le \frac{2}{3} \text{ (B)} \\ 1, & \frac{2}{3} < x \le 1 \text{ (C)} \end{cases}$$

In every region, $P(\omega_1 \mid x) + P(\omega_2 \mid x) = 1$.

**(b)** Sketch or plot $P(\omega_1 \mid x)$ and $P(\omega_2 \mid x)$ versus $x$.
Label the three regions $A$, $B$, and $C$. (3 points)

**(c)** Write the Bayes decision rule for minimizing the probability of error:

$$
\text{Decide } \omega_1 \text{ if } P(\omega_1 \mid x) > P(\omega_2 \mid x), \text{ else decide } \omega_2.
$$


The decision rule should give the ranges of $x$ in which the classifier decides
$\omega_1$ and in which it decides $\omega_2$, with the precise location of the boundary where the classifier switches. (3 points)

$$\text{Decide } \omega_1 \text{ if } 0 \le x < \tfrac{1}{2}, \qquad \text{decide } \omega_2 \text{ if } \tfrac{1}{2} < x \le 1$$

**(d)** Compute the minimum achievable probability of error. (3 points)

$$P(\text{error}) = \int_0^1 \min\big[P(\omega_1 \mid x),\, P(\omega_2 \mid x)\big]\, p(x)\, dx$$

- Regions A and C: the smaller posterior is 0, so they contribute 0 error.
- Region B, $\frac{1}{3} \le x < \frac{1}{2}$: we decide $\omega_1$, so the error is $P(\omega_2 \mid x) = 3x - 1$. This is a triangle with width $\frac{1}{6}$ and height $\frac{1}{2}$, so the area is $\frac{1}{2} \cdot \frac{1}{6} \cdot \frac{1}{2} = \frac{1}{24}$.
- Region B, $\frac{1}{2} < x \le \frac{2}{3}$: $\frac{1}{24}$ by symmetry.

$$P(\text{error}) = 0 + \frac{1}{24} + \frac{1}{24} + 0 = \frac{1}{12} \approx 0.083$$

This is the Bayes error. It is nonzero because both classes produce values of $x$ in Region B, so even the optimal classifier will sometimes be wrong there.


```python
#B
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 1000)

# Posteriors by region (from part a)
P1 = np.piecewise(x, [x < 1/3, (x >= 1/3) & (x <= 2/3), x > 2/3],
                  [1, lambda x: 2 - 3*x, 0])
P2 = np.piecewise(x, [x < 1/3, (x >= 1/3) & (x <= 2/3), x > 2/3],
                  [0, lambda x: 3*x - 1, 1])

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(x, P1, label=r'$P(\omega_1 \mid x)$', linewidth=2)
ax.plot(x, P2, label=r'$P(\omega_2 \mid x)$', linewidth=2)

# Region boundaries and labels
for b in [1/3, 2/3]:
    ax.axvline(b, color='gray', linestyle='--', linewidth=1)
for xc, name in [(1/6, 'A'), (1/2, 'B'), (5/6, 'C')]:
    ax.text(xc, 1.08, f'Region {name}', ha='center', fontsize=11)

ax.set_xlim(0, 1)
ax.set_ylim(-0.05, 1.15)
ax.set_xlabel('$x$')
ax.set_ylabel('Posterior probability')
ax.set_title('Posterior probabilities vs. $x$')
ax.legend(loc='center right')
plt.show()
```

## Problem 4: Optimal Classification with Three Classes

An optimal classifier minimizes the probability of error by choosing the class with the largest posterior
probability $P(\omega_i \mid x)$.  Usually we are not handed the posteriors; we know (or estimate) each
class's likelihood $p(x \mid \omega_i)$ and prior $P(\omega_i)$, and obtain the posteriors from Bayes' rule.

Consider a one-dimensional feature $x \in [0,1]$ with three classes $\omega_1, \omega_2, \omega_3$ and priors
$$
P(\omega_1) = \tfrac{1}{4}, \qquad P(\omega_2) = \tfrac{1}{2}, \qquad P(\omega_3) = \tfrac{1}{4}.
$$
The class-conditional densities (likelihoods), each zero outside $[0,1]$, are
$$
p(x \mid \omega_1) = 2(1-x), \qquad
p(x \mid \omega_2) =
\begin{cases}
4x, & 0 \le x \le \tfrac{1}{2} \\
4(1-x), & \tfrac{1}{2} < x \le 1
\end{cases}, \qquad
p(x \mid \omega_3) = 2x.
$$

**(a)** Verify that each likelihood integrates to 1.  Then find the evidence $p(x)$ using the law of
total probability, and use Bayes' rule to find the posteriors $P(\omega_1 \mid x)$, $P(\omega_2 \mid x)$,
and $P(\omega_3 \mid x)$ on $0 \le x \le \tfrac{1}{2}$ and on $\tfrac{1}{2} < x \le 1$. (4 points)

- $p(x \mid \omega_1) = 2(1-x)$: goes from 2 down to 0, area $= \frac{1}{2} \cdot 1 \cdot 2 = 1$
- $p(x \mid \omega_2)$: goes from 0 up to 2 at $x = \frac{1}{2}$, then back down to 0, area $= \frac{1}{2} \cdot 1 \cdot 2 = 1$
- $p(x \mid \omega_3) = 2x$: goes from 0 up to 2, area $= \frac{1}{2} \cdot 1 \cdot 2 = 1$

$$p(x) = \tfrac{1}{4}\,p(x \mid \omega_1) + \tfrac{1}{2}\,p(x \mid \omega_2) + \tfrac{1}{4}\,p(x \mid \omega_3)$$

The $\omega_1$ and $\omega_3$ terms combine: $\frac{1}{4} \cdot 2(1-x) + \frac{1}{4} \cdot 2x = \frac{1}{2}$, so

$$p(x) = \begin{cases} \frac{1}{2} + 2x, & 0 \le x \le \frac{1}{2} \\ \frac{5}{2} - 2x, & \frac{1}{2} < x \le 1 \end{cases}$$

*Posteriors* (Bayes' rule): $P(\omega_i \mid x) = \dfrac{p(x \mid \omega_i)P(\omega_i)}{p(x)}$

On $0 \le x \le \frac{1}{2}$:

$$P(\omega_1 \mid x) = \frac{1-x}{1+4x}, \quad P(\omega_2 \mid x) = \frac{4x}{1+4x}, \quad P(\omega_3 \mid x) = \frac{x}{1+4x}$$

On $\frac{1}{2} < x \le 1$:

$$P(\omega_1 \mid x) = \frac{1-x}{5-4x}, \quad P(\omega_2 \mid x) = \frac{4-4x}{5-4x}, \quad P(\omega_3 \mid x) = \frac{x}{5-4x}$$

On each interval the numerators add up to the denominator, so the posteriors sum to 1.

**(b)** Sketch or plot the three posteriors over $x \in [0,1]$.
Label the regions where each class attains the largest posterior. (3 points)

**(c)** Write the Bayes decision rule that minimizes probability of error and determine the decision regions
explicitly; i.e., find the boundary point(s) where the predicted class changes. (3 points)

Since all posteriors at a given $x$ share the denominator $p(x)$, compare the numerators $p(x \mid \omega_i)P(\omega_i)$.

On $[0, \frac{1}{2}]$: $\frac{1}{2}x < 2x$, so $\omega_3$ never wins. Setting $\omega_1 = \omega_2$:

$$\tfrac{1}{2}(1-x) = 2x \implies x = \tfrac{1}{5}$$

On $(\frac{1}{2}, 1]$: $\frac{1}{2}(1-x) < 2 - 2x$, so $\omega_1$ never wins. Setting $\omega_2 = \omega_3$:

$$2 - 2x = \tfrac{1}{2}x \implies x = \tfrac{4}{5}$$

Bayes decision rule:

$$\text{Decide } \begin{cases} \omega_1, & 0 \le x < \frac{1}{5} \\ \omega_2, & \frac{1}{5} < x < \frac{4}{5} \\ \omega_3, & \frac{4}{5} < x \le 1 \end{cases}$$

**(d)** Compute the minimum achievable probability of error. (3 points)

$$P(\text{error}) = 1 - \int_0^1 \max_i\, p(x \mid \omega_i)P(\omega_i)\, dx$$

$$\int_0^{1/5} \tfrac{1}{2}(1-x)\,dx = \tfrac{9}{100}, \quad \int_{1/5}^{1/2} 2x\,dx = \tfrac{21}{100}, \quad \int_{1/2}^{4/5} (2-2x)\,dx = \tfrac{21}{100}, \quad \int_{4/5}^{1} \tfrac{1}{2}x\,dx = \tfrac{9}{100}$$

$$P(\text{error}) = 1 - \left(\tfrac{9}{100} + \tfrac{21}{100} + \tfrac{21}{100} + \tfrac{9}{100}\right) = 1 - 0.6 = 0.4$$

**(e)** Repeat (c) and (d) with equal priors $P(\omega_1)=P(\omega_2)=P(\omega_3)=\tfrac{1}{3}$.
How do the priors move the decision boundaries? (3 points)

*Hint:* The probability of error at a given $x$ is $1 - \max_i P(\omega_i \mid x)$, so
$$
P(\text{error}) = \int_0^1 \left(1 - \max_i P(\omega_i \mid x)\right) p(x)\, dx
= 1 - \int_0^1 \max_i\, p(x \mid \omega_i)\, P(\omega_i)\, dx.
$$
\
With equal priors, every numerator is multiplied by the same $\frac{1}{3}$, so we compare the likelihoods directly.

On $[0, \frac{1}{2}]$: $2x < 4x$, so $\omega_3$ never wins. $2(1-x) = 4x \implies x = \frac{1}{3}$

On $(\frac{1}{2}, 1]$: $2(1-x) < 4(1-x)$, so $\omega_1$ never wins. $4(1-x) = 2x \implies x = \frac{2}{3}$

$$\text{Decide } \begin{cases} \omega_1, & 0 \le x < \frac{1}{3} \\ \omega_2, & \frac{1}{3} < x < \frac{2}{3} \\ \omega_3, & \frac{2}{3} < x \le 1 \end{cases}$$

$$\int_0^{1/3} 2(1-x)\,dx = \tfrac{5}{9}, \qquad \int_{1/3}^{1/2} 4x\,dx = \tfrac{5}{18}$$

By symmetry the other two regions give the same values, so

$$P(\text{error}) = 1 - \tfrac{1}{3} \cdot 2\left(\tfrac{5}{9} + \tfrac{5}{18}\right) = 1 - \tfrac{5}{9} = \tfrac{4}{9} \approx 0.444$$

Increasing $P(\omega_2)$ from $\frac{1}{3}$ to $\frac{1}{2}$ moves the boundaries outward ($\frac{1}{3} \to \frac{1}{5}$, $\frac{2}{3} \to \frac{4}{5}$), expanding $\omega_2$'s region, and lowers $P(\text{error})$ from $\frac{4}{9}$ to $0.4$.

```python
#4b
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 1000)
left = x <= 0.5

P1 = np.where(left, (1 - x) / (1 + 4*x), (1 - x) / (5 - 4*x))
P2 = np.where(left, 4*x / (1 + 4*x), (4 - 4*x) / (5 - 4*x))
P3 = np.where(left, x / (1 + 4*x), x / (5 - 4*x))

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(x, P1, label=r'$P(\omega_1 \mid x)$', linewidth=2)
ax.plot(x, P2, label=r'$P(\omega_2 \mid x)$', linewidth=2)
ax.plot(x, P3, label=r'$P(\omega_3 \mid x)$', linewidth=2)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.05)
ax.set_xlabel('$x$')
ax.set_ylabel('Posterior probability')
ax.set_title('Posteriors with priors (1/4, 1/2, 1/4)')
ax.legend()
plt.show()
```

## Problem 5: Risk


Automatic Content Recognition (ACR) aims to identify content such as TV
shows, movies, or sports broadcasts. In this scenario, we treat each piece of
content as a class in a multiclass classification problem, where there are
$K$ known pieces of content. 

During distribution, the same content may be re-encoded multiple times to adapt
to available bandwidth or loss characteristics, and may be subject to "visual
enhancements" by devices like set-top boxes or TVs. These modifications
introduce noise, but the model must still correctly classify the content.

Because video and audio are composed of continuous streams of frames or audio
samples, from which we generate feature vectors $\mathbf{x}(t)$ at different
time points $t$, we are not forced to make an immediate decision. Instead, the
ACR system can choose to *reject* an observation as unrecognizable, rather than
risk misclassifying it. When the cost of rejection is low, it may be preferable
to reject an observation and wait for more data to improve confidence.

Let $\lambda$ denote our loss function.

\begin{equation}
\lambda(\alpha_i| \omega_j) =
\begin{cases}
  0 & i = j \text{ for } i, j = 1, ..., K \\
  \lambda_r & i = K + 1 \quad (\text{rejection}) \\
  \lambda_s & i \neq j \text{ for } i \le K \quad (\text{substitution}) 
\end{cases}
\end{equation}

where 

* $\lambda_r$ is the loss incurred for rejecting the observation
  (i.e., deciding to wait).  Let action $(K+1)$ denote rejection.
* $\lambda_s$ is the loss incurred for a substitution error (i.e.,
  classifying the observation as the wrong content).


**(a)** At decision time, the ACR system must either choose to classify the content into one of the known classes or reject the observation (i.e., postpone the decision). Find the risk function for the case where the system chooses to classify the observation (i.e., no rejection), $R(\alpha_i | \mathbf{x}, \text{choose})$. The risk function represents the expected loss conditioned on the observation $\mathbf{x}$ and the decision to classify. (3 points)

$$R(\alpha_i \mid \mathbf{x}, \text{choose}) = \sum_{j \ne i} \lambda_S\, P(\omega_j \mid \mathbf{x}) = \lambda_S \big(1 - P(\omega_i \mid \mathbf{x})\big)$$

The conditional risk of action $\alpha_i$ is the expected loss over the true class:

$$R(\alpha_i \mid \mathbf{x}) = \sum_{j=1}^{K} \lambda(\alpha_i \mid \omega_j)\, P(\omega_j \mid \mathbf{x})$$

For classifying as $\omega_i$ ($i \le K$), the loss is 0 when $j = i$ and $\lambda_s$ when $j \ne i$:

$$R(\alpha_i \mid \mathbf{x}, \text{choose}) = \sum_{j \ne i} \lambda_s\, P(\omega_j \mid \mathbf{x}) = \lambda_s \big(1 - P(\omega_i \mid \mathbf{x})\big)$$

using $\sum_{j=1}^{K} P(\omega_j \mid \mathbf{x}) = 1$.


**(b)** If the risk of classifying the content (i.e., choosing) exceeds the
risk of rejecting the observation, then choosing will increase the overall
risk. To minimize the risk, we should only choose to classify if:

$$R(\alpha_i|\mathbf{x}, \text{choose}) \leq R(\alpha_i|\mathbf{x}, \text{reject})$$

Using this criterion, show that the minimum risk is obtained by classifying
the observation as class $\omega_i$ if:

$$P(\omega_i|\mathbf{x}) \geq P(\omega_j|\mathbf{x}) \text{ for all } j \neq i$$
which means choosing the class with the highest posterior probability.
Furthermore, show that we should classify the observation as $\omega_i$ if

$$P(\omega_i | \mathbf{x}) \geq 1 - \frac{\lambda_r}{\lambda_s}, $$

and reject otherwise. (5 points)

*Which class to choose:* $\lambda_s$ is the same for every class, so minimizing $\lambda_s\big(1 - P(\omega_i \mid \mathbf{x})\big)$ is equivalent to maximizing $P(\omega_i \mid \mathbf{x})$. The minimum risk is therefore obtained by choosing $\omega_i$ where

$$P(\omega_i \mid \mathbf{x}) \ge P(\omega_j \mid \mathbf{x}) \quad \text{for all } j \ne i$$

*Choose or reject:* rejection costs $\lambda_r$ regardless of the true class:

$$R(\alpha_{K+1} \mid \mathbf{x}) = \sum_{j=1}^{K} \lambda_r\, P(\omega_j \mid \mathbf{x}) = \lambda_r \sum_{j=1}^{K} P(\omega_j \mid \mathbf{x}) = \lambda_r$$

Classify only if $R(\alpha_i \mid \mathbf{x}, \text{choose}) \le R(\alpha_i \mid \mathbf{x}, \text{reject})$:

$$\lambda_s \big(1 - P(\omega_i \mid \mathbf{x})\big) \le \lambda_r$$

$$1 - P(\omega_i \mid \mathbf{x}) \le \frac{\lambda_r}{\lambda_s}$$

$$P(\omega_i \mid \mathbf{x}) \ge 1 - \frac{\lambda_r}{\lambda_s}$$

Classify as $\omega_i$ if this holds; otherwise reject.


**(c)** What happens to the decisions made when $\lambda_r = 0$? (2 points)

If $\lambda_r = 0$, the threshold becomes $1 - 0 = 1$, so the system classifies only when $P(\omega_i \mid \mathbf{x}) = 1$. Otherwise it rejects everything: waiting costs nothing, so there is no reason to risk a wrong guess.



**(d)** What happens to the decisions made when $\lambda_r > \lambda_s$? (2 points)

If $\lambda_r > \lambda_s$, then $\frac{\lambda_r}{\lambda_s} > 1$ and the threshold $1 - \frac{\lambda_r}{\lambda_s} < 0$. Every posterior is $\ge 0$, so the condition always holds and the system never rejects. It always classifies with the highest-posterior class, because a wrong guess costs less than rejecting.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

X0 = np.random.uniform(0, 2, size=(100, 2))
X1 = np.random.uniform(5, 7, size=(100, 2))
y0 = np.zeros(100)
y1 = np.ones(100)

X_syn = np.vstack([X0, X1])
y_syn = np.concatenate([y0, y1])

print(X_syn.shape, y_syn.shape)
fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(X_syn[y_syn == 0, 0], X_syn[y_syn == 0, 1], label='Class 0', alpha=0.7)
ax.scatter(X_syn[y_syn == 1, 0], X_syn[y_syn == 1, 1], label='Class 1', alpha=0.7)
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_title('Synthetic dataset: two disjoint squares')
ax.legend()
ax.set_aspect('equal')
plt.show()
```

```python
def gda_fit(X, y):
    phi = np.mean(y == 1)
    mu0 = np.mean(X[y == 0], axis=0)
    mu1 = np.mean(X[y == 1], axis=0)

    D0 = X[y == 0] - mu0
    D1 = X[y == 1] - mu1
    D = np.vstack([D0, D1])
    Sigma = (1 / len(X)) * D.T @ D

    return phi, mu0, mu1, Sigma

phi, mu0, mu1, Sigma = gda_fit(X_syn, y_syn)
print("phi =", phi)
print("mu0 =", mu0)
print("mu1 =", mu1)
print("Sigma =\n", Sigma)
```

```python
def gda_predict(X, phi, mu0, mu1, Sigma):
    Sigma_inv = np.linalg.inv(Sigma)
    w = Sigma_inv @ (mu1 - mu0)
    b = (-0.5 * (mu1 @ Sigma_inv @ mu1 - mu0 @ Sigma_inv @ mu0)
         + np.log(phi / (1 - phi)))
    scores = X @ w + b
    return (scores > 0).astype(int)  # 1 where score > 0, else 0
def test_gda_perfect_accuracy():
        phi, mu0, mu1, Sigma = gda_fit(X_syn, y_syn)
        y_pred = gda_predict(X_syn, phi, mu0, mu1, Sigma)
        accuracy = np.mean(y_pred == y_syn)
        assert accuracy == 1.0, f"Expected 100% accuracy, got {accuracy}"
        print("Test passed: accuracy =", accuracy)

test_gda_perfect_accuracy()
```
